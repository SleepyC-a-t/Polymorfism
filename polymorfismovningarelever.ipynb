### Övning 1 — Repetition: override ger polymorfism

Loopa igenom en lista med minst en av varje (inklusive ett vanligt `Instrument`-objekt) och anropa `spela()` på alla.

```python
# Skriv din kod här
class Instrument:
    def __init__(self,name):
        self.name = name

    def play(self):
        print(self.name, "is playing")

class Guitar(Instrument):
    def play(self):
        print(self.name, "is playing it's strings")

class Drums(Instrument):
    def play(self):
        print(self.name, "is drumming")

instruments = [Instrument("Instrument"), Guitar("Guitar"), Drums("Drums")]
for i in instruments:
    i.play()
```

### Övning 4 — Polymorfism via en gemensam funktion

Skapa `Instrument`, `Gitarr` och `Trumma` (som i övning 1). Skriv en funktion `spela_instrument(instrument)` som bara anropar `instrument.spela()`.

Anropa funktionen med ett `Gitarr`-objekt och ett `Trumma`-objekt.

```python
# Skriv din kod här
class Instrument:
    def __init__(self,name):
        self.name = name

    def play(self):
        print(self.name, "is playing")

class Guitar(Instrument):
    def play(self):
        print(self.name, "is playing it's strings")

class Drums(Instrument):
    def play(self):
        print(self.name, "is drumming")

def play_instrument(instrument):
    instrument.play()

instrument1 = Guitar("Guitar")
instrument2 = Drums("Drums")
play_instrument(instrument1)
play_instrument(instrument2)
```

### Övning 5 — Duck typing

Skapa två klasser, `Telefon` och `Dörrklocka`, som INTE ärver från någon gemensam superklass. Ge båda en metod `ring()` som skriver ut ett lämpligt meddelande (t.ex. `"Telefonen ringer!"` respektive `"Dörrklockan ringer!"`).

Loopa igenom en lista med ett objekt av vardera klass och anropa `ring()` på båda.

```python
# Skriv din kod här
class Telephone:
    def ring():
        print("The telephone is ringing")

class Doorbell:
    def ring():
        print("The doorbell is ringing")

ringing_objects = [Telephone, Doorbell]
for x in ringing_objects:
    x.ring()
```
