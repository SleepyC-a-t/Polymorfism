### Övning 1 — Repetition: override ger polymorfism

Skapa en klass `Instrument` med attributet `namn` och en metod `spela()` som skriver ut `"<namn> spelas."`

Skapa `Gitarr(Instrument)` som overridar `spela()` till `"<namn> knäpper strängarna."`, och `Trumma(Instrument)` som overridar den till `"<namn> trummar."`

Loopa igenom en lista med minst en av varje (inklusive ett vanligt `Instrument`-objekt) och anropa `spela()` på alla.

```python
# Skriv din kod här
class Instrument:
    def __init__(self,name):
        self.name = name

    def play(self):
        print(self.name, "is playing")

class Guitar(Instrument):
    def play(self):
        print(self.name, "is playing it's strings")

class Drums(Instrument):
    def play(self):
        print(self.name, "is drumming")

instruments = [Instrument("Instrument"), Guitar("Guitar"), Drums("Drums")]
for i in instruments:
    i.play()
```

### Övning 4 — Polymorfism via en gemensam funktion

Skapa `Instrument`, `Gitarr` och `Trumma` (som i övning 1). Skriv en funktion `spela_instrument(instrument)` som bara anropar `instrument.spela()`.

Anropa funktionen med ett `Gitarr`-objekt och ett `Trumma`-objekt.

```python
# Skriv din kod här
class Instrument:
    def __init__(self,name):
        self.name = name

    def play(self):
        print(self.name, "is playing")

class Guitar(Instrument):
    def play(self):
        print(self.name, "is playing it's strings")

class Drums(Instrument):
    def play(self):
        print(self.name, "is drumming")

def play_instrument(instrument):
    instrument.play()

instrument1 = Guitar("Guitar")
instrument2 = Drums("Drums")
play_instrument(instrument1)
play_instrument(instrument2)
```

### Övning 7 — Överlagra `__add__`

Skapa en klass `Poang` med attributet `varde`. Lägg till en `__add__`-metod så att två `Poang`-objekt kan adderas ihop med `+` (resultatet ska vara ett nytt `Poang`-objekt med summan av värdena).

Skapa två `Poang`-objekt, addera dem, och skriv ut resultatets `varde`.

```python
# Skriv din kod här
class Points:
    def __init__(self, value):
        self.value = value

    def __add__(self, other_points):
        return Points(self.value + other_points.value)


point_sum = Points(10) + Points(15)
print(point_sum.value)
```

### Övning 9 — `__str__` är också polymorfism

Skapa två klasser, `Biljett` och `Kvitto`, båda med en egen `__str__`-metod som returnerar en läsbar textrad (fritt vad de innehåller, t.ex. biljettens resmål respektive kvittots totalsumma).

Skriv ut ett objekt av vardera klass i en loop med `print()`.

```python
# Skriv din kod här
class Ticket:
    def __init__(self, destination, price):
        self.destination = destination
        self.price = price

    def __str__(self):
        return "Destination: " + self.destination + " Ticket price: " + self.price

class Receipt:
    def __init__(self, product, price):
        self.product = product
        self.price = price

    def __str__(self):
        return "Product: " + self.product + " Product price: " + self.price

object1 = Receipt("Book", "100kr")
object2 = Ticket("Spain", "1200kr")
listt = [object1, object2]

for item in listt:
    print(item)
```

### Övning 10 — Sammanfattande övning: blandat band

Skapa `Instrument`, `Gitarr` och `Trumma` (som i övning 1). Skapa också en duck-typad klass `Radio` (som i övning 6) med en `spela()`-metod, men som INTE ärver från `Instrument`.

Skapa en lista `bandet` med minst en `Gitarr`, en `Trumma` och en `Radio`. Loopa igenom listan, anropa `spela()` på varje, och använd `isinstance()` för att skriva ut om objektet är ett "riktigt" instrument eller inte.

```python
# Skriv din kod här
class Instrument:
    def __init__(self,name):
        self.name = name

    def play(self):
        print(self.name, "is playing")

class Guitar(Instrument):
    def play(self):
        print(self.name, "is playing it's strings")

class Drums(Instrument):
    def play(self):
        print(self.name, "is drumming")

class Radio:
    def __init__(self, name):
        self.name = name

    def play(self):
        print(self.name, "is playing music")

guitar1 = Guitar("Guitar")
drums1 = Drums("Drums")
radio1 = Radio("Radio")
band = [guitar1, drums1, radio1] #Can make the list with a more efficient method but this works for now
for i in band:
    if isinstance(i, Instrument):
        print(i.name, "is a true instrument")
    else:
        print(i.name, "is not a true instrument")
```
